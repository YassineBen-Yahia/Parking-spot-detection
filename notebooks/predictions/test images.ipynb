```python
from ultralytics import YOLO
import random
from glob import glob

model=YOLO('yolo_train/exp/weights/best.pt')

val_images_dir = 'data/split/images/test'
image_paths = glob(val_images_dir + '/*')
sample_images = random.sample(image_paths, min(3, len(image_paths)))

# Predict and show results
results = model.predict(sample_images, save=False, conf=0.25)

for r in results:
    r.show()
```

```python
import cv2
from ultralytics import YOLO

# Load trained model
model = YOLO("yolo_train/exp/weights/best.pt")

# Open video file (or use 0 for webcam)
cap = cv2.VideoCapture("video.mp4")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLO prediction on the frame
    results = model.predict(frame, conf=0.25, verbose=False)

    # YOLO gives annotated frame (with bboxes & labels)
    annotated_frame = results[0].plot()

    # Show frame in a window
    cv2.imshow("YOLO Predictions", annotated_frame)

    # Press 'q' to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

cv2.destroyAllWindows()
```
