```python
from ultralytics import YOLO
import random
from glob import glob

model=YOLO('yolo_train/exp/weights/best.pt')

val_images_dir = 'data/split/images/test'
image_paths = glob(val_images_dir + '/*')
sample_images = random.sample(image_paths, min(3, len(image_paths)))

# Predict and show results
results = model.predict(sample_images, save=False, conf=0.25)

for r in results:
    r.show()
```
