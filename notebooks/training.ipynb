```python
import os
from ultralytics import YOLO

# Paths
project_dir = 'yolo_train_results'  # Directory to save all results
os.makedirs(project_dir, exist_ok=True)

data_yaml = 'data/yolo_data.yaml'  # Path to your YOLOv8 data config file
# You may need to create this YAML file to point to your split/train/val/test folders

# Load a YOLOv8 model (use 'yolov8n.pt' for nano, 'yolov8s.pt' for small, etc.)
model = YOLO('yolov8n.pt')

# Train
results = model.train(
    data=data_yaml,
    epochs=50,
    imgsz=640,
    project=project_dir,
    name='exp',
    save=True,
    val=True,
    exist_ok=True
)

# Evaluate on test set and save results
metrics = model.val(data=data_yaml, split='test', project=project_dir, name='exp_eval', exist_ok=True)

# Save metrics to a file
import json
with open(os.path.join(project_dir, 'exp_eval', 'metrics.json'), 'w') as f:
    json.dump(metrics.results_dict, f, indent=2)

print('Training and evaluation complete. All results saved in:', project_dir)
```
