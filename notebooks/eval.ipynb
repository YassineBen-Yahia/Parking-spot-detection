```python
import os
import json
from ultralytics import YOLO



model_path = 'yolo_train/exp/weights/best.pt'
eval_dir = 'yolo_train/exp_eval'

# Load the model
model = YOLO(model_path)

# Load metrics from JSON if available
metrics_path = os.path.join(eval_dir, 'metrics.json')
if os.path.exists(metrics_path):
    with open(metrics_path, 'r') as f:
        metrics = json.load(f)
    print('Loaded metrics from file:')
    for k, v in metrics.items():
        print(f'{k}: {v}')
else:
    print('metrics.json not found. Run evaluation and save metrics first.')


results = model.val(data='data/split/data.yaml', split='test', project='yolo_train', name='exp_eval', exist_ok=True)

print(f"mAP@0.5: {results.box.map50:.4f}")
print(f"mAP@0.5:0.95: {results.box.map:.4f}")

# Show all available metrics
for k, v in results.results_dict.items():
    print(f'{k}: {v}')
```
